```python
import manim as mn
from manim import *

config.media_width = "75%"
config.background_color = "#11111b"
config.verbosity = "WARNING"

print(mn.__version__)
```

```python
unit = 0.2
```

```python
%%manim -ql Intro

class Intro(Scene):
    def construct(self):      
        plus_symb = MathTex("+", font_size=256, color=WHITE)
        times_symb = MathTex(r"\times", font_size=256, color=WHITE)
        times_symb.rotate(270 * DEGREES)
        
        self.wait()
        self.play(Write(times_symb))
        self.wait()
        self.play(times_symb.animate.rotate(45 * DEGREES))
        self.play(Transform(times_symb, plus_symb))
        self.wait()
        self.play(FadeOut(times_symb))
```

```python
%%manim -qh DotsCreation

class DotsCreation(Scene):
    def construct(self):      
        red_dot = Dot(color=RED)
        blue_dot = Dot(color=BLUE)
        white_dot = Dot(color=WHITE)
        
        red_grid = VGroup(*[red_dot.copy() for _ in range(10)])
        blue_grid = VGroup(*[blue_dot.copy() for _ in range(7)])
        
        red_grid.arrange_in_grid(rows=1, buff=unit)
        blue_grid.arrange_in_grid(rows=1, buff=unit)
        
        self.play(DrawBorderThenFill(red_grid))
        self.play(red_grid.animate.to_edge(LEFT))
        self.wait(1/2)
        self.play(DrawBorderThenFill(blue_grid))
        self.play(blue_grid.animate.to_edge(RIGHT))
        self.wait()

        self.play(blue_grid.animate.next_to(red_grid, RIGHT).set_color(RED))
        self.wait()
        self.play(blue_grid.animate.to_edge(RIGHT).set_color(BLUE))
        self.play(red_grid.animate.next_to(blue_grid, LEFT).set_color(BLUE))
        self.wait(2)
        self.play(red_grid.animate.to_edge(LEFT).set_color(RED))
        self.wait()
        self.play(red_grid.animate.move_to(ORIGIN).set_color(WHITE), blue_grid.animate.set_opacity(0))

        animations = []
        for dot in red_grid:
            animations.append(Transform(dot, white_dot))
        self.play(*animations)
        
        self.wait(5)
```

```python
%%manim -ql DiscreteLine

class DiscreteLine(Scene):
    def construct(self):
        white_dot = Dot(color=WHITE)
        self.add(white_dot)
        self.wait()
```
