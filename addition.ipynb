```python
import manim as mn
from manim import *

config.media_width = "75%"
config.background_color = "#11111b"
config.verbosity = "WARNING"

print(mn.__version__)
```

```python
unit = 0.2
right_motion = 2 * RIGHT * unit
left_motion = 2 * LEFT * unit
```

```python
%%manim -qm Intro

class Intro(Scene):
    def construct(self):      
        plus_symb = MathTex("+", font_size=256, color=WHITE)
        times_symb = MathTex(r"\times", font_size=256, color=WHITE)
        times_symb.rotate(270 * DEGREES)
        
        self.wait()
        self.play(Write(times_symb))
        self.wait()
        self.play(times_symb.animate.rotate(45 * DEGREES))
        self.play(Transform(times_symb, plus_symb))
        self.wait()
        self.play(FadeOut(times_symb))
```

```python
%%manim -qh DotsCreation

class DotsCreation(Scene):
    def construct(self):      
        red_dot = Dot(color=RED)
        blue_dot = Dot(color=BLUE)
        white_dot = Dot(color=WHITE)
        
        red_grid = VGroup(*[red_dot.copy() for _ in range(10)])
        blue_grid = VGroup(*[blue_dot.copy() for _ in range(7)])
        
        red_grid.arrange_in_grid(rows=1, buff=unit)
        blue_grid.arrange_in_grid(rows=1, buff=unit)
        
        self.play(DrawBorderThenFill(red_grid))
        self.play(red_grid.animate.to_edge(LEFT))
        self.wait(1/2)
        self.play(DrawBorderThenFill(blue_grid))
        self.play(blue_grid.animate.to_edge(RIGHT))
        self.wait()

        self.play(blue_grid.animate.next_to(red_grid, RIGHT).set_color(RED))
        self.wait()
        self.play(blue_grid.animate.to_edge(RIGHT).set_color(BLUE))
        self.play(red_grid.animate.next_to(blue_grid, LEFT).set_color(BLUE))
        self.wait(2)
        self.play(red_grid.animate.to_edge(LEFT).set_color(RED))
        self.wait()
        self.play(red_grid.animate.move_to(ORIGIN).set_color(WHITE), blue_grid.animate.set_opacity(0))

        animations = []
        for dot in red_grid:
            animations.append(Transform(dot, white_dot))
        self.play(*animations)
        
        self.wait(5)
```

```python
%%manim -qm DiscreteLine

class DiscreteLine(Scene):
    def construct(self):
        dot_amount = 18
        white_dot = Dot(color=WHITE)
        natural_dots = VGroup(*(white_dot.copy() for i in range(dot_amount)))
        self.add(natural_dots)

        for i in range(dot_amount - 1):
            animation = []
            for j, dot in zip(range(dot_amount), natural_dots):
                if i < j:
                    animation.append(dot.animate.shift(2 * RIGHT * unit))
            self.play(*animation, run_time=.75)
            
        self.wait()
        
        original_dots = natural_dots.copy()
        original_dots.set_fill(opacity=0.2)
        original_dots.set_stroke(color=WHITE, width=1)
        for dot in original_dots:
            dot.scale(0.8)
            
        natural_dots.set_z_index(1)
        original_dots.set_z_index(0)

        self.play(natural_dots[3].animate.set_color(GREEN))
        self.add(original_dots)
        self.play(natural_dots.animate.shift(3 * right_motion))
        self.wait()

        self.play(natural_dots.animate.shift(3 * left_motion))
        self.wait()
        self.play(natural_dots[3].animate.set_color(WHITE))

        self.remove(original_dots)
        
        zero_dot = Dot(color=BLACK, stroke_width=1.5, stroke_color=WHITE)
        self.play(natural_dots.animate.shift(right_motion)) 
        natural_dots.insert(0, zero_dot)
        self.play(Create(zero_dot))
        self.wait()
        natural_dots[5].set_color(RED)
        original_dots[5].set_color(RED)
        self.add(original_dots)
        self.play(natural_dots.animate.shift(right_motion * 5))
        self.wait()
        self.play(natural_dots.animate.shift(left_motion * 5))
        natural_dots[5].set_color(WHITE)
        original_dots[5].set_color(WHITE)
        self.wait()
        self.play(natural_dots.animate.shift(right_motion * 3))
        self.wait()
        self.play(natural_dots.animate.shift(right_motion * 2))
        self.wait()
        self.play(natural_dots.animate.shift(right_motion * 4))
        self.wait()
        self.play(natural_dots.animate.shift(right_motion * 7))
        self.wait()
        self.play(natural_dots.animate.shift(right_motion * 2))
        self.play(*[FadeOut(mob) for mob in self.mobjects])
```

```python
%%manim -ql Recap

class Recap(Scene):
    def construct(self):
        crab = SVGMobject("media/cuddlyferris.svg")
        crabs = VGroup(*[crab.copy() for _ in range(3)])
        crabs.arrange(RIGHT, buff=6 * unit)

        self.play(Write(crabs))
        self.wait()

        self.play(*[
            Transform(
                c,
                Dot(
                    point=c.get_center(),
                    color=WHITE,
                    stroke_width=1.5,
                    stroke_color=WHITE,
                )
            )
            for c in crabs
        ])

        self.wait()

        # Les trois points se rapprochent
        self.play(
            crabs.animate.arrange(RIGHT, buff=2 * unit),
            run_time=2
        )

        self.wait()
```

```python
%%manim -ql FullLine

class FullLine(Scene):
    def construct(self):
        integers = VGroup(*[Dot(color=WHITE, stroke_width=1.5, stroke_color=WHITE) for _ in range(49)])
        pseudo_int = VGroup(*[Dot(color=WHITE, stroke_width=1.5, stroke_color=WHITE) for _ in range(45)])
        integers.arrange_in_grid(rows=1, buf=(2 * unit))
        pseudo_int.arrange_in_grid(rows=1, buf=(2 * unit))
        integers[24].set_color(BLACK).set_stroke(width=1.5, color=WHITE)
        pseudo_int[22].set_color(BLACK).set_stroke(width=1.5, color=WHITE)
        self.add(integers)
        self.wait()
        values = [ 3, 4, -2, -5, 7, -1, -1, -4, -6]
        for value in values:
            self.play(integers.animate.shift(right_motion * value))
            self.wait(0.3)
        self.play(integers.animate.shift(left_motion * sum(values)))
        self.wait()
        self.add(pseudo_int)
        self.play(integers.animate.shift(DOWN))
        self.wait()
        self.play(integers.animate.shift(right_motion))
```
